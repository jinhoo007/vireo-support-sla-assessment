```python
# Cell 4: Load datasets

tickets = pd.read_csv("tickets.csv")
agents = pd.read_csv("agents.csv")
orders = pd.read_csv("orders.csv")
customers = pd.read_csv("customers.csv")
products = pd.read_csv("products.csv")

print("Datasets loaded successfully.\n")

print(f"tickets   : {tickets.shape[0]:,} rows × {tickets.shape[1]} columns")
print(f"agents    : {agents.shape[0]:,} rows × {agents.shape[1]} columns")
print(f"orders    : {orders.shape[0]:,} rows × {orders.shape[1]} columns")
print(f"customers : {customers.shape[0]:,} rows × {customers.shape[1]} columns")
print(f"products  : {products.shape[0]:,} rows × {products.shape[1]} columns")
```

```python
# Cell 5: Validate required ticket columns

required_ticket_columns = {
    "ticket_id",
    "created_at",
    "first_response_at",
    "resolved_at",
    "status",
    "channel",
    "customer_id",
    "order_id",
    "product_sku",
    "category",
    "priority",
    "assigned_team",
    "agent_id",
    "transfers",
    "csat_score",
    "refund_amount_inr",
    "refund_reason_code",
    "replacement_issued",
    "customer_message",
    "agent_notes",
    "source_system"
}

actual_ticket_columns = set(tickets.columns)

missing_ticket_columns = required_ticket_columns - actual_ticket_columns

if missing_ticket_columns:
    raise ValueError(
        f" Missing ticket columns: {sorted(missing_ticket_columns)}"
    )

print("All required ticket columns are present.")

print("\nTicket columns:")
for i, column in enumerate(tickets.columns, start=1):
    print(f"{i:2}. {column}")
```

```python
# Cell 6: Inspect raw tickets

print("First 5 ticket records:")
display(tickets.head())

print("\nData types:")
display(tickets.dtypes.to_frame("dtype"))

print("\nTicket status distribution:")
display(
    tickets["status"]
    .value_counts(dropna=False)
    .rename_axis("status")
    .reset_index(name="tickets")
)

print("\nChannel distribution:")
display(
    tickets["channel"]
    .value_counts(dropna=False)
    .rename_axis("channel")
    .reset_index(name="tickets")
)
```

```python
# Cell 7: Duplicate ticket audit

duplicate_mask = tickets["ticket_id"].duplicated(keep=False)

duplicate_rows = tickets.loc[duplicate_mask].copy()

duplicate_ticket_ids = duplicate_rows["ticket_id"].nunique()

print(f"Total raw rows              : {len(tickets):,}")
print(f"Unique ticket IDs           : {tickets['ticket_id'].nunique():,}")
print(f"Rows in duplicated ticket IDs: {len(duplicate_rows):,}")
print(f"Duplicated ticket IDs       : {duplicate_ticket_ids:,}")

print("\nDuplicate source-system combinations:")
display(
    duplicate_rows
    .groupby(["ticket_id", "source_system"])
    .size()
    .reset_index(name="rows")
    .head(20)
)
```

```python
# Cell 8: Inspect duplicate migration records

duplicate_source_counts = (
    duplicate_rows
    .groupby("ticket_id")["source_system"]
    .agg(lambda x: sorted(set(x)))
    .reset_index(name="source_systems")
)

print("How many duplicated tickets contain both source systems?")

both_systems = duplicate_source_counts[
    duplicate_source_counts["source_systems"].apply(
        lambda x: set(x) == {"helpdesk", "legacy_fd"}
    )
]

print(f"Duplicated IDs with helpdesk + legacy_fd: {len(both_systems):,}")

print("\nExample duplicated ticket:")
if len(both_systems) > 0:
    example_ticket_id = both_systems.iloc[0]["ticket_id"]

    display(
        tickets[
            tickets["ticket_id"] == example_ticket_id
        ].sort_values("source_system")
    )
```

```python
# Cell 9: Deduplicate tickets

tickets_clean = tickets.copy()

# Prefer the current helpdesk record over the legacy migration record.
source_priority = {
    "helpdesk": 0,
    "legacy_fd": 1
}

tickets_clean["_source_priority"] = (
    tickets_clean["source_system"]
    .map(source_priority)
    .fillna(99)
)

tickets_clean = (
    tickets_clean
    .sort_values(
        by=["ticket_id", "_source_priority"],
        kind="stable"
    )
    .drop_duplicates(
        subset="ticket_id",
        keep="first"
    )
    .drop(columns="_source_priority")
    .reset_index(drop=True)
)

print("Canonical ticket dataset created.\n")

print(f"Raw ticket rows       : {len(tickets):,}")
print(f"Clean ticket rows     : {len(tickets_clean):,}")
print(f"Rows removed          : {len(tickets) - len(tickets_clean):,}")
print(f"Unique ticket IDs     : {tickets_clean['ticket_id'].nunique():,}")
```

```python
# Cell 10: Convert UTC timestamps to IST

timestamp_columns = [
    "created_at",
    "first_response_at",
    "resolved_at"
]

for column in timestamp_columns:
    # Parse the CSV timestamp explicitly as UTC.
    tickets_clean[column] = pd.to_datetime(
        tickets_clean[column],
        errors="coerce",
        utc=True
    )

    # Convert UTC → Asia/Kolkata (IST).
    tickets_clean[f"{column}_ist"] = (
        tickets_clean[column]
        .dt.tz_convert("Asia/Kolkata")
    )

print("UTC → IST conversion completed.\n")

display(
    tickets_clean[
        [
            "ticket_id",
            "created_at",
            "created_at_ist",
            "first_response_at",
            "first_response_at_ist",
            "resolved_at",
            "resolved_at_ist"
        ]
    ].head(10)
)
```

```python
# CELL 11 — Timestamp quality checks

timestamp_check = pd.DataFrame({
    "column": timestamp_columns,
    "missing_values": [
        tickets_clean[col].isna().sum()
        for col in timestamp_columns
    ],
    "invalid_or_missing_pct": [
        round(tickets_clean[col].isna().mean() * 100, 2)
        for col in timestamp_columns
    ]
})

print("Timestamp quality check:")
display(timestamp_check)

# First response cannot be before ticket creation
invalid_response_order = (
    tickets_clean["first_response_at"] < tickets_clean["created_at"]
).sum()

# Resolution cannot be before ticket creation
invalid_resolution_order = (
    tickets_clean["resolved_at"] < tickets_clean["created_at"]
).sum()

print(f"\nFirst response before creation : {invalid_response_order:,}")
print(f"Resolution before creation    : {invalid_resolution_order:,}")

if invalid_response_order > 0 or invalid_resolution_order > 0:
    raise ValueError("❌ Invalid timestamp ordering detected.")

print("Timestamp ordering checks passed.")
```

```python
# CELL 12 — Calculate first-response time

tickets_clean["response_time_minutes"] = (
    tickets_clean["first_response_at"] - tickets_clean["created_at"]
).dt.total_seconds() / 60

# Basic quality checks
negative_response_times = (
    tickets_clean["response_time_minutes"] < 0
).sum()

missing_response_times = (
    tickets_clean["response_time_minutes"].isna()
).sum()

print(f"Negative response times : {negative_response_times:,}")
print(f"Missing response times  : {missing_response_times:,}")

if negative_response_times > 0:
    raise ValueError("Negative response times detected.")

if missing_response_times > 0:
    raise ValueError("Missing response times detected.")

print("\nResponse-time summary (minutes):")
display(
    tickets_clean["response_time_minutes"]
    .describe()
    .to_frame("minutes")
)

print(" Response-time calculation completed.")
```

```python
# CELL 13 — Apply channel-specific SLA targets

sla_minutes = {
    "chat": 15,
    "voice": 120,
    "social": 240,
    "email": 480
}

tickets_clean["sla_target_minutes"] = (
    tickets_clean["channel"]
    .str.lower()
    .map(sla_minutes)
)

unknown_channels = tickets_clean.loc[
    tickets_clean["sla_target_minutes"].isna(),
    "channel"
].dropna().unique()

if len(unknown_channels) > 0:
    raise ValueError(
        f"Unknown channel(s) found: {sorted(unknown_channels.tolist())}"
    )

print("SLA targets applied successfully:\n")

sla_summary = (
    tickets_clean[["channel", "sla_target_minutes"]]
    .drop_duplicates()
    .sort_values("channel")
)

display(sla_summary)

print(" Channel-specific SLA targets applied.")
```

```python
# CELL 14 — Determine SLA breach

tickets_clean["sla_breach"] = (
    tickets_clean["response_time_minutes"]
    > tickets_clean["sla_target_minutes"]
)

tickets_clean["sla_status"] = np.where(
    tickets_clean["sla_breach"],
    "BREACH",
    "PASS"
)

print("Overall SLA status:")
display(
    tickets_clean["sla_status"]
    .value_counts()
    .rename_axis("sla_status")
    .reset_index(name="tickets")
)

print("\nBreach rate:")
overall_breach_rate = tickets_clean["sla_breach"].mean() * 100
print(f"{overall_breach_rate:.2f}%")

print("SLA breach calculation completed.")
```

```python
# CELL 15 — Calculate SLA credit exposure

SLA_CREDIT_PER_BREACH = 350

tickets_clean["sla_credit_exposure_inr"] = np.where(
    tickets_clean["sla_breach"],
    SLA_CREDIT_PER_BREACH,
    0
)

total_breaches = int(tickets_clean["sla_breach"].sum())
total_credit_exposure = int(
    tickets_clean["sla_credit_exposure_inr"].sum()
)

print(f"Total tickets analyzed : {len(tickets_clean):,}")
print(f"Total SLA breaches     : {total_breaches:,}")
print(f"Breach rate            : {overall_breach_rate:.2f}%")
print(f"SLA credit exposure    : ₹{total_credit_exposure:,.0f}")

print("\nExposure check:")
print(
    f"{total_breaches:,} breaches × ₹{SLA_CREDIT_PER_BREACH:,} "
    f"= ₹{total_credit_exposure:,.0f}"
)

print("SLA financial exposure calculated.")
```

```python
# CELL 16 — Validate and prepare agent roster

agents_clean = agents.copy()

# Clean column names
agents_clean.columns = agents_clean.columns.str.strip()

# Required columns from the actual agents.csv
required_agent_columns = {
    "agent_id",
    "name",
    "site",
    "team",
    "shift",
    "tier",
    "from_date",
    "to_date"
}

missing_agent_columns = sorted(
    required_agent_columns - set(agents_clean.columns)
)

if missing_agent_columns:
    raise ValueError(
        f"Missing agent columns: {missing_agent_columns}"
    )

# Convert roster dates
agents_clean["from_date"] = pd.to_datetime(
    agents_clean["from_date"],
    errors="coerce"
)

agents_clean["to_date"] = pd.to_datetime(
    agents_clean["to_date"],
    errors="coerce"
)

# Validate dates
invalid_date_rows = (
    agents_clean["from_date"].isna()
).sum()

if invalid_date_rows > 0:
    raise ValueError(
        f" {invalid_date_rows:,} roster rows have invalid from_date values."
    )

print("Agent roster validation completed.\n")

print(f"Assignment rows : {len(agents_clean):,}")
print(f"Unique agents   : {agents_clean['agent_id'].nunique():,}")

print("\nShift distribution:")
display(
    agents_clean["shift"]
    .value_counts(dropna=False)
    .rename_axis("shift")
    .reset_index(name="assignments")
)

print("\nTier distribution:")
display(
    agents_clean["tier"]
    .value_counts(dropna=False)
    .rename_axis("tier")
    .reset_index(name="assignments")
)

print("\nRoster preview:")
display(agents_clean.head(10))
```

```python
# CELL 17 — Attach historical roster assignment

# Ticket creation date in IST
tickets_clean["_created_date_ist"] = (
    tickets_clean["created_at_ist"]
    .dt.tz_localize(None)
    .dt.normalize()
)

# Prepare roster
agents_lookup = agents_clean.copy()

# Open-ended assignments remain active indefinitely
agents_lookup["to_date_filled"] = agents_lookup["to_date"].fillna(
    pd.Timestamp("2099-12-31")
)

# Rename roster columns to avoid conflicts
agents_lookup = agents_lookup.rename(columns={
    "name": "roster_agent_name",
    "site": "roster_site",
    "team": "roster_team",
    "shift": "roster_shift",
    "tier": "roster_tier"
})

# Match tickets to agent assignments
ticket_roster = tickets_clean.merge(
    agents_lookup[
        [
            "agent_id",
            "roster_agent_name",
            "roster_site",
            "roster_team",
            "roster_shift",
            "roster_tier",
            "from_date",
            "to_date_filled"
        ]
    ],
    on="agent_id",
    how="left"
)

# Keep the assignment active when the ticket was created
active_assignment = (
    (ticket_roster["_created_date_ist"] >= ticket_roster["from_date"]) &
    (ticket_roster["_created_date_ist"] <= ticket_roster["to_date_filled"])
)

ticket_roster = ticket_roster.loc[active_assignment].copy()

# Check matching quality
roster_match_counts = (
    ticket_roster.groupby("ticket_id")
    .size()
)

multiple_roster_matches = (
    roster_match_counts > 1
).sum()

matched_ticket_ids = set(ticket_roster["ticket_id"])
all_ticket_ids = set(tickets_clean["ticket_id"])

unmatched_ticket_ids = (
    all_ticket_ids - matched_ticket_ids
)

print("Historical roster matching results:")
print(f"Tickets in clean dataset       : {len(all_ticket_ids):,}")
print(f"Tickets matched to roster      : {len(matched_ticket_ids):,}")
print(f"Tickets without roster match   : {len(unmatched_ticket_ids):,}")
print(f"Tickets with multiple matches  : {multiple_roster_matches:,}")

if multiple_roster_matches > 0:
    print("\nMultiple roster assignments detected.")
else:
    print("\nNo multiple roster assignments detected.")
```

```python
# CELL 18 — Build final analysis dataset

analysis = ticket_roster.copy()

# Shift from historical roster
analysis["agent_shift"] = analysis["roster_shift"]

# Operational shift based on ticket creation time in IST
def determine_shift(hour):
    if 6 <= hour < 14:
        return "Morning"
    elif 14 <= hour < 22:
        return "Day"
    else:
        return "Night"

analysis["operational_shift"] = (
    analysis["created_at_ist"]
    .dt.hour
    .apply(determine_shift)
)

# Weekly reporting period
analysis["week_start"] = (
    analysis["created_at_ist"]
    .dt.tz_localize(None)
    .dt.to_period("W-SUN")
    .dt.start_time
)

print("Final analysis dataset created.")
print(f"Rows: {len(analysis):,}")

print("\nAgent shift distribution:")
display(
    analysis["agent_shift"]
    .value_counts(dropna=False)
    .rename_axis("agent_shift")
    .reset_index(name="tickets")
)

print("\nOperational shift distribution:")
display(
    analysis["operational_shift"]
    .value_counts(dropna=False)
    .rename_axis("operational_shift")
    .reset_index(name="tickets")
)
```

```python
# CELL 19 — Weekly shift SLA report

weekly_shift_report = (
    analysis
    .groupby(
        ["week_start", "agent_shift"],
        dropna=False
    )
    .agg(
        tickets=("ticket_id", "nunique"),
        breaches=("sla_breach", "sum"),
        sla_credit_exposure_inr=("sla_credit_exposure_inr", "sum")
    )
    .reset_index()
)

# Calculate breach rate
weekly_shift_report["breach_rate_pct"] = (
    weekly_shift_report["breaches"]
    / weekly_shift_report["tickets"]
    * 100
)

# Avoid division-by-zero issues
weekly_shift_report["breach_rate_pct"] = (
    weekly_shift_report["breach_rate_pct"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

# Sort for easier reading
weekly_shift_report = weekly_shift_report.sort_values(
    ["week_start", "breaches"],
    ascending=[True, False]
).reset_index(drop=True)

print("Weekly shift SLA report created.")
print(f"Reporting weeks: {weekly_shift_report['week_start'].nunique():,}")

display(weekly_shift_report.head(30))
```

```python
# CELL 20 — Weekly agent SLA report

weekly_agent_report = (
    analysis
    .groupby(
        [
            "week_start",
            "agent_id",
            "roster_agent_name",
            "agent_shift",
            "roster_site",
            "roster_team",
            "roster_tier"
        ],
        dropna=False
    )
    .agg(
        tickets=("ticket_id", "nunique"),
        breaches=("sla_breach", "sum"),
        sla_credit_exposure_inr=("sla_credit_exposure_inr", "sum")
    )
    .reset_index()
)

# Calculate breach rate
weekly_agent_report["breach_rate_pct"] = (
    weekly_agent_report["breaches"]
    / weekly_agent_report["tickets"]
    * 100
)

# Protect against division-by-zero
weekly_agent_report["breach_rate_pct"] = (
    weekly_agent_report["breach_rate_pct"]
    .replace([np.inf, -np.inf], np.nan)
    .fillna(0)
)

# Sort:
# 1. Week
# 2. Highest breach volume
# 3. Highest breach rate
weekly_agent_report = weekly_agent_report.sort_values(
    [
        "week_start",
        "breaches",
        "breach_rate_pct"
    ],
    ascending=[True, False, False]
).reset_index(drop=True)

print(" Weekly agent SLA report created.")
print(f"Reporting weeks : {weekly_agent_report['week_start'].nunique():,}")
print(f"Agent-week rows  : {len(weekly_agent_report):,}")

display(weekly_agent_report.head(50))
```

```python
# CELL 21 — Overall SLA validation

print("=" * 70)
print("OVERALL SLA VALIDATION")
print("=" * 70)

total_tickets = analysis["ticket_id"].nunique()
total_breaches = int(analysis["sla_breach"].sum())
total_passes = total_tickets - total_breaches

calculated_breach_rate = (
    total_breaches / total_tickets * 100
    if total_tickets > 0 else 0
)

calculated_exposure = int(
    analysis["sla_credit_exposure_inr"].sum()
)

expected_exposure = total_breaches * SLA_CREDIT_PER_BREACH

print(f"Total tickets       : {total_tickets:,}")
print(f"SLA passes          : {total_passes:,}")
print(f"SLA breaches        : {total_breaches:,}")
print(f"Breach rate         : {calculated_breach_rate:.2f}%")
print(f"Credit exposure     : ₹{calculated_exposure:,.0f}")
print(f"Expected exposure   : ₹{expected_exposure:,.0f}")

# Validation checks
assert total_tickets == total_passes + total_breaches
assert calculated_exposure == expected_exposure

print("\nTicket count reconciliation passed.")
print(" ₹350 exposure reconciliation passed.")
print(" Overall SLA validation passed.")
```

```python
# CELL 22 — Validate channel SLA targets

expected_sla = {
    "chat": 15,
    "voice": 120,
    "social": 240,
    "email": 480
}

channel_validation = (
    analysis
    .groupby("channel")
    .agg(
        tickets=("ticket_id", "nunique"),
        breaches=("sla_breach", "sum"),
        avg_response_minutes=("response_time_minutes", "mean"),
        median_response_minutes=("response_time_minutes", "median")
    )
    .reset_index()
)

channel_validation["sla_target_minutes"] = (
    channel_validation["channel"]
    .str.lower()
    .map(expected_sla)
)

channel_validation["breach_rate_pct"] = (
    channel_validation["breaches"]
    / channel_validation["tickets"]
    * 100
)

channel_validation = channel_validation.sort_values(
    "breach_rate_pct",
    ascending=False
).reset_index(drop=True)

print("Channel-level SLA performance:")
display(channel_validation)

# Verify every channel has the expected target
for channel, target in expected_sla.items():
    actual_targets = (
        analysis.loc[
            analysis["channel"].str.lower() == channel,
            "sla_target_minutes"
        ]
        .dropna()
        .unique()
    )

    assert len(actual_targets) == 1
    assert actual_targets[0] == target

print("Channel SLA targets validated.")
```

```python
# CELL 23 — Shift-level performance

shift_report = (
    analysis
    .groupby("agent_shift", dropna=False)
    .agg(
        tickets=("ticket_id", "nunique"),
        breaches=("sla_breach", "sum"),
        avg_response_minutes=("response_time_minutes", "mean"),
        median_response_minutes=("response_time_minutes", "median"),
        sla_credit_exposure_inr=("sla_credit_exposure_inr", "sum")
    )
    .reset_index()
)

shift_report["breach_rate_pct"] = (
    shift_report["breaches"]
    / shift_report["tickets"]
    * 100
)

shift_report["share_of_all_breaches_pct"] = (
    shift_report["breaches"]
    / total_breaches
    * 100
)

shift_report = shift_report.sort_values(
    "breaches",
    ascending=False
).reset_index(drop=True)

print("Overall shift performance:")
display(shift_report)

print("\nHighest breach-volume shift:")
top_volume_shift = shift_report.iloc[0]

print(
    f"{top_volume_shift['agent_shift']} — "
    f"{int(top_volume_shift['breaches']):,} breaches "
    f"({top_volume_shift['share_of_all_breaches_pct']:.2f}% of all breaches)"
)

print("Shift analysis completed.")
```

```python
# CELL 24 — Test the Morning-shift hypothesis

morning_row = shift_report[
    shift_report["agent_shift"].astype(str).str.lower() == "morning"
]

if len(morning_row) == 0:
    raise ValueError("❌ Morning shift was not found in the roster.")

morning_breaches = int(morning_row.iloc[0]["breaches"])
morning_breach_share = float(
    morning_row.iloc[0]["share_of_all_breaches_pct"]
)

morning_is_highest_volume = (
    shift_report.iloc[0]["agent_shift"] == morning_row.iloc[0]["agent_shift"]
)

print("Morning-shift hypothesis:")
print(f"Morning breaches          : {morning_breaches:,}")
print(f"Share of all breaches     : {morning_breach_share:.2f}%")
print(f"Highest breach volume     : {morning_is_highest_volume}")

if morning_is_highest_volume:
    print(
        "\n Neha's claim is supported by breach volume."
    )
else:
    print(
        "\nNeha's claim is NOT supported by breach volume."
    )

print("\nImportant:")
print("Breach volume and breach rate are different measures.")
print("We should report both rather than declaring a shift 'worst' from volume alone.")
```

```python
# CELL 37 — AI classification schema

PRIMARY_ISSUES = [
    "Delivery & Shipping",
    "Billing & Payments",
    "Connectivity",
    "Returns & Refunds",
    "Charging & Battery",
    "App & Firmware",
    "Audio Quality",
    "Product Enquiry",
    "Warranty & Repair",
    "Account & Login",
    "Other"
]

COMPLEXITY_LEVELS = [
    "Low",
    "Medium",
    "High"
]

OPERATIONAL_FACTORS = [
    "Queue Backlog",
    "Customer Information Missing",
    "Technical Troubleshooting",
    "Cross-Team Dependency",
    "Transfer / Handoff",
    "Policy / Approval",
    "Delivery / Logistics Dependency",
    "No Clear Operational Factor",
    "Other"
]

BREACH_DRIVER_VALUES = [
    "Yes",
    "No",
    "Unclear"
]

print("Classification schema loaded.")
```

```python
# CELL 38 — Prompt builder and JSON validation

def build_classification_prompt(customer_message, agent_notes):
    return f"""
You are analyzing a customer-support ticket for an operations team.

Classify the ticket using ONLY the allowed values below.

Return ONLY a JSON object.
Do not include markdown.
Do not include explanations.

JSON schema:
{{
  "primary_issue": "...",
  "complexity": "...",
  "operational_factor": "...",
  "likely_breach_driver": "..."
}}

Allowed primary_issue:
{", ".join(PRIMARY_ISSUES)}

Allowed complexity:
{", ".join(COMPLEXITY_LEVELS)}

Allowed operational_factor:
{", ".join(OPERATIONAL_FACTORS)}

Allowed likely_breach_driver:
{", ".join(BREACH_DRIVER_VALUES)}

Important:
- Do not assume that a ticket caused an SLA breach merely because it actually breached.
- Use the ticket text and agent notes only.
- If the operational cause cannot reasonably be inferred, use "Unclear" or "No Clear Operational Factor".

Customer message:
{customer_message}

Agent notes:
{agent_notes}
"""


def validate_classification(result):
    if not isinstance(result, dict):
        return False, "Output is not a JSON object"

    required = [
        "primary_issue",
        "complexity",
        "operational_factor",
        "likely_breach_driver"
    ]

    missing = [x for x in required if x not in result]

    if missing:
        return False, f"Missing fields: {missing}"

    if result["primary_issue"] not in PRIMARY_ISSUES:
        return False, "Invalid primary_issue"

    if result["complexity"] not in COMPLEXITY_LEVELS:
        return False, "Invalid complexity"

    if result["operational_factor"] not in OPERATIONAL_FACTORS:
        return False, "Invalid operational_factor"

    if result["likely_breach_driver"] not in BREACH_DRIVER_VALUES:
        return False, "Invalid likely_breach_driver"

    return True, "Valid"
```

```python
# CELL 39 — Gemini ticket classifier

def classify_ticket_gemini(
    customer_message,
    agent_notes,
    max_retries=2
):
    prompt = build_classification_prompt(
        customer_message,
        agent_notes
    )

    for attempt in range(max_retries + 1):

        try:
            response = client.models.generate_content(
                model="gemini-2.5-flash",
                contents=prompt,
                config={
                    "temperature": 0.1,
                    "response_mime_type": "application/json"
                }
            )

            raw_text = response.text.strip()

            result = json.loads(raw_text)

            valid, validation_message = validate_classification(result)

            if valid:
                return {
                    "success": True,
                    "result": result,
                    "error": None,
                    "attempts": attempt + 1
                }

            error = validation_message

        except Exception as e:
            error = str(e)

        if attempt < max_retries:
            time.sleep((2 ** attempt) + random.uniform(0, 1))

    return {
        "success": False,
        "result": None,
        "error": error,
        "attempts": max_retries + 1
    }

print("Gemini classifier ready.")
```

```python
# CELL 49 — Generate Final Weekly SLA Reports

print("=" * 70)
print("GENERATING FINAL WEEKLY SLA REPORTS")
print("=" * 70)


# ============================================================
# 1. WEEKLY AGENT-LEVEL REPORT
# ============================================================

weekly_agent_report = (
    analysis
    .groupby(
        [
            "week_start",
            "agent_id",
            "roster_agent_name",
            "operational_shift"
        ]
    )
    .agg(
        tickets=("ticket_id", "count"),
        sla_breaches=("sla_breach", "sum")
    )
    .reset_index()
)

# Calculate breach rate
weekly_agent_report["breach_rate"] = (
    weekly_agent_report["sla_breaches"]
    / weekly_agent_report["tickets"]
    * 100
)

# Calculate automatic store-credit exposure
weekly_agent_report["credit_exposure_inr"] = (
    weekly_agent_report["sla_breaches"] * 350
)

# Rename columns for clean business output
weekly_agent_report = weekly_agent_report.rename(
    columns={
        "roster_agent_name": "agent_name",
        "operational_shift": "shift"
    }
)

# Sort chronologically and then by breach volume
weekly_agent_report = (
    weekly_agent_report
    .sort_values(
        ["week_start", "sla_breaches", "breach_rate"],
        ascending=[True, False, False]
    )
    .reset_index(drop=True)
)


# ============================================================
# 2. WEEKLY SHIFT-LEVEL REPORT
# ============================================================

weekly_shift_report = (
    analysis
    .groupby(
        [
            "week_start",
            "operational_shift"
        ]
    )
    .agg(
        tickets=("ticket_id", "count"),
        sla_breaches=("sla_breach", "sum")
    )
    .reset_index()
)

# Calculate breach rate
weekly_shift_report["breach_rate"] = (
    weekly_shift_report["sla_breaches"]
    / weekly_shift_report["tickets"]
    * 100
)

# Calculate automatic store-credit exposure
weekly_shift_report["credit_exposure_inr"] = (
    weekly_shift_report["sla_breaches"] * 350
)

# Rename for business readability
weekly_shift_report = weekly_shift_report.rename(
    columns={
        "operational_shift": "shift"
    }
)

# Sort chronologically
weekly_shift_report = (
    weekly_shift_report
    .sort_values(
        ["week_start", "shift"]
    )
    .reset_index(drop=True)
)


# ============================================================
# 3. FORMAT REPORT COLUMNS
# ============================================================

weekly_agent_report["week_start"] = pd.to_datetime(
    weekly_agent_report["week_start"]
).dt.strftime("%Y-%m-%d")

weekly_shift_report["week_start"] = pd.to_datetime(
    weekly_shift_report["week_start"]
).dt.strftime("%Y-%m-%d")


# ============================================================
# 4. SAVE REPORTS
# ============================================================

weekly_agent_report.to_csv(
    "weekly_sla_report.csv",
    index=False
)

weekly_shift_report.to_csv(
    "weekly_shift_report.csv",
    index=False
)


# ============================================================
# 5. VALIDATION
# ============================================================

print("\nWEEKLY AGENT REPORT")
print("-" * 70)

print(
    f"Rows generated : "
    f"{len(weekly_agent_report):,}"
)

print(
    f"Columns        : "
    f"{list(weekly_agent_report.columns)}"
)

display(
    weekly_agent_report.head(10)
)


print("\nWEEKLY SHIFT REPORT")
print("-" * 70)

print(
    f"Rows generated : "
    f"{len(weekly_shift_report):,}"
)

print(
    f"Columns        : "
    f"{list(weekly_shift_report.columns)}"
)

display(
    weekly_shift_report.head(10)
)


# ============================================================
# 6. REPORT RECONCILIATION
# ============================================================

agent_total_tickets = (
    weekly_agent_report["tickets"].sum()
)

agent_total_breaches = (
    weekly_agent_report["sla_breaches"].sum()
)

shift_total_tickets = (
    weekly_shift_report["tickets"].sum()
)

shift_total_breaches = (
    weekly_shift_report["sla_breaches"].sum()
)

print("\nREPORT RECONCILIATION")
print("-" * 70)

print(
    f"Source tickets           : {total_tickets:,}"
)

print(
    f"Agent report tickets     : {agent_total_tickets:,}"
)

print(
    f"Shift report tickets     : {shift_total_tickets:,}"
)

print(
    f"Source breaches          : {total_breaches:,}"
)

print(
    f"Agent report breaches    : {agent_total_breaches:,}"
)

print(
    f"Shift report breaches    : {shift_total_breaches:,}"
)

print(
    f"Agent report reconciles  : "
    f"{agent_total_tickets == total_tickets and agent_total_breaches == total_breaches}"
)

print(
    f"Shift report reconciles  : "
    f"{shift_total_tickets == total_tickets and shift_total_breaches == total_breaches}"
)


# ============================================================
# 7. FINAL FILE CONFIRMATION
# ============================================================

print("\nFILES CREATED")
print("-" * 70)

print("✓ weekly_sla_report.csv")
print("✓ weekly_shift_report.csv")

print("\n" + "=" * 70)
print("WEEKLY SLA REPORT GENERATION COMPLETE")
print("=" * 70)
```
